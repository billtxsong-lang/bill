***The problem I want to address is whether the server owner can determine how long the currently online players will continue to play, in order to decide whether to issue more permits later.***

To solve the problem, I would like to draw a histogram to show among all the time periods, at what time is the number of concurrent logins the highest. Additionally, I also created a regression model to predict the player's online duration based on the time of their account activation. This can help the server owner predict how long the players who are currently online will remain active, facilitating better management of the server. 

So I created a histogram and I measure it by the hour, and then determine which hour during the day has the largest number of simultaneous online users. If the login and logout times are not at the exact hour, they will be calculated as if they were at the next whole hour. For example, if a player logs in at 11:37 and logs out at 13:59, the entire 3-hour period from 11:00 to 13:00 will be counted as the player's online time.

```r
data <- sessions |>
  mutate(
    start_time = str_trim(start_time),
    end_time = str_trim(end_time),
    start_time = as.POSIXct(start_time, format = "%d/%m/%Y %H:%M", tz = "UTC"),
    end_time = as.POSIXct(end_time, format = "%d/%m/%Y %H:%M", tz = "UTC")
  ) |>
  filter(!is.na(start_time), !is.na(end_time))

events <- data |>
  pivot_longer(c(start_time, end_time), names_to = "event_type", values_to = "time") |>
  mutate(change = if_else(event_type == "start_time", 1, -1)) |>
  arrange(time) |>
  mutate(concurrent = cumsum(change))

hourly_online <- events |>
  mutate(hour = format(floor_date(time, "hour"), "%H")) |>
  group_by(hour) |>
  summarise(avg_online = mean(concurrent)) |>
  mutate(hour = as.numeric(hour))

options(repr.plot.width=8, repr.plot.height=4)

ggplot(hourly_online, aes(x = hour, y = avg_online)) +
  geom_col(fill = "darkorange") +
  labs(
    title = "Average Concurrent Users by Hour (Event-based)",
    x = "Hour of Day",
    y = "Relative Number of Concurrent Users"
  )+
scale_y_continuous(labels = NULL)




```

From this graph, it can be seen that this is not a normal distribution. The largest number of people was at 3 o'clock. And the number remained high even from 20 o'clock. However, starting from 8 o'clock, the number of people suddenly decreased, and by 11 o'clock, there were no people online at all. So from 2 o'clock to 5 o'clock, the server owner can issue more permit. From 11 to 12 o'clock, there is basically no need to distribute the permit.

**Why is this method appropriate?**

KNN regression is good because it is a non-parametric method. This means we don't need to assume our data follows a specific distribution. It can handle non-linear relationships well and is simple to understand. It looks for similar data points to make predictions, which works nicely when we don't know the exact pattern in our data.

**Which assumptions are required, if any, to apply the method selected??**

KNN needs only a few simple assumptions. First, similar inputs should give similar outputs. Second, we must scale all features to the same range because distance calculation is important. Third, the data should be clean without too many outliers. Lastly, the number of features should not be too large compared to our data size.

**What are the potential limitations or weaknesses of the method selected?**

KNN has several weaknesses. It is slow because it must check all training data for each prediction. It doesn't work well with high-dimensional data. The choice of k value is very sensitive - too small causes overfitting, too large causes underfitting. It also requires storing all training data in memory, which can be expensive.

**How are you going to compare and select the model?**

I will use cross-validation to test different k values. For each k, we calculate RMSE and MAE scores. I repeat this process multiple times and choose the k with the best average performance. I will also compare KNN with simple models like linear regression to make sure KNN is truly better.

**How will you process the data?**

First, I split data into 80% training and 20% testing sets. We do this before any processing to avoid data leakage. Then I do 5-fold cross-validation on the training set only. I must scale features using training data and apply the same scaling to test data. The final model is trained on all training data and evaluated only once on the test set.
